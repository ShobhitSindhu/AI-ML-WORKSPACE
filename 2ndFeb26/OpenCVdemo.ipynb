```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("960px-Nietzsche187c.jpg")   

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

blur = cv2.GaussianBlur(gray, (5, 5), 0)

edges = cv2.Canny(blur, 100, 200)



plt.figure(figsize=(12,6))

plt.subplot(1,4,1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(1,4,2)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale")
plt.axis("off")

plt.subplot(1,4,3)
plt.imshow(blur, cmap='gray')
plt.title("Blur")
plt.axis("off")

plt.subplot(1,4,4)
plt.imshow(edges, cmap='gray')
plt.title("Edges")
plt.axis("off")

plt.show()

```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")
results = model("table_chair.jpg", show=True)

chair_count = 0

for r in results:
    for box in r.boxes:
        class_name = r.names[int(box.cls)]
        confidence = float(box.conf)

        if class_name == "chair":
            chair_count += 1
            print(
                "Class:", class_name,
                "| Confidence:", round(confidence, 2),
                "| Coordinates:", box.xyxy.tolist()  # x1, y1, x2, y2
            )

print("\nTotal number of chairs detected:", chair_count)
```
